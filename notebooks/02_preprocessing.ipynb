```python
import pandas as pd
import sys
sys.path.append('../src')
from preprocess import preprocess

# Charger le dataset
df = pd.read_csv('../data/ChatGPT.csv', encoding='latin-1', low_memory=False)
# Tester sur quelques exemples
samples = df['Tweet'].dropna().head(5).tolist()
for tweet in samples:
    print("ORIGINAL:", tweet[:100])
    print("CLEANED :", preprocess(tweet))
    print("---")
```

```python
from tqdm.notebook import tqdm
tqdm.pandas()

print("Applying preprocessing pipeline...")
df['cleaned_tweet'] = df['Tweet'].progress_apply(preprocess)

df = df[df['cleaned_tweet'].str.strip() != '']
df = df.dropna(subset=['cleaned_tweet'])

print(f"\nDataset size after cleaning: {len(df)}")
df[['Tweet', 'cleaned_tweet']].head(5)
```

```python
df.to_csv('../data/ChatGPT_cleaned.csv', index=False)
print("Saved to data/ChatGPT_cleaned.csv ✓")
```
